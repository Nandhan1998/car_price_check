```python
import pandas as pd
import numpy as np
veh_data=pd.read_csv("C:\\Users\\aruna\\Desktop\\ML_vehicle_price_detection\\auto.txt",na_values="?")
veh_data.head()
```

```python
veh_data.isnull().sum()
```

```python
# remove the unwanted columns
veh_data=veh_data.drop("normalized-losses",axis=1)
veh_data.head(2)
```

```python
veh_data.columns
```

```python
#  checking for the columns data type and converting into the int/float
veh_data["price"].describe()
```

```python
mak={'alfa-romero':1, 'audi':2, 'bmw':3, 'chevrolet':4, 'dodge':5, 'honda':6,'isuzu':7, 'jaguar':8, 'mazda':9, 'mercedes-benz':10, 'mercury':11,
       'mitsubishi':12, 'nissan':13, 'peugot':14, 'plymouth':15, 'porsche':16, 'renault':17,
       'saab':18, 'subaru':19, 'toyota':20, 'volkswagen':21, 'volvo':22}
veh_data["make"].replace(mak,inplace=True)

fueltype={'gas':1, 'diesel':2}
veh_data["fuel-type"].replace(fueltype,inplace=True)

aspirati={'std':1, 'turbo':2}
veh_data["aspiration"].replace(aspirati,inplace=True)

numofdoors={'two':2, 'four':4}
veh_data["num-of-doors"].replace(numofdoors,inplace=True)

bodystyle={'convertible':1, 'hatchback':2 ,'sedan':3, 'wagon':4, 'hardtop':5}
veh_data["body-style"].replace(bodystyle,inplace=True)

drivewheels={'rwd':1, 'fwd':2, '4wd':3}
veh_data["drive-wheels"].replace(drivewheels,inplace=True)

enginelocation={'front':1, 'rear':2}
veh_data["engine-location"].replace(enginelocation,inplace=True)

enginetype={'dohc':1, 'ohcv':2, 'ohc':3, 'l':4, 'rotor':5, 'ohcf':6, 'dohcv':7}
veh_data["engine-type"].replace(enginetype,inplace=True)

numofcylinders={'four':4, 'six':6, 'five':5, 'three':3, 'twelve':12, 'two':2, 'eight':8}
veh_data["num-of-cylinders"].replace(numofcylinders,inplace=True)

fuelsystem={'mpfi':1, '2bbl':2, 'mfi':3, '1bbl':4, 'spfi':5, '4bbl':6, 'idi':7, 'spdi':8}
veh_data["fuel-system"].replace(fuelsystem,inplace=True)
```

```python
veh_data.isnull().sum()
```

```python
# from sklearn.impute import SimpleImputer
# fill_nan=SimpleImputer(missing_values=np.nan,strategy="mean")
# num_cols = ['bore', 'stroke', 'horsepower', 'peak-rpm', 'price','num-of-doors']
# veh_data[num_cols] =fill_nan.fit_transform(veh_data[num_cols])
veh_data=veh_data.dropna()
```

```python
veh_data.isnull().sum()
```

```python
# ## create the traing and testing data set
```

```python
# ### drop the cloums that are the values are less then the coefficent 
veh_data=veh_data.drop("city-mpg",axis=1)
veh_data=veh_data.drop("peak-rpm",axis=1)
veh_data=veh_data.drop("horsepower",axis=1)
veh_data=veh_data.drop("curb-weight",axis=1)
```

```python
from sklearn.model_selection import train_test_split
x=veh_data.drop("price",axis=1)
y=veh_data["price"]
xtrain,xtest,ytrain,ytest=train_test_split(x,y,test_size=0.2,random_state=0)
```

```python
from sklearn.linear_model import LinearRegression
liner_model=LinearRegression()
liner_model.fit(xtrain,ytrain)
```

```python
# check the accuracy of the training set
liner_model.score(xtrain,ytrain)
```

```python
# Check for the accuracy of the testing 
liner_model.score(xtest,ytest)
```

```python
# liner_model.coef_
```

```python
predictor=xtrain.columns
coef=pd.Series(liner_model.coef_,predictor).sort_values()
print(coef)
```

```python
# #### test the algorithms
import math
y_predict=liner_model.predict(xtest)
# print(y_predict)
# find  the mean square error
from sklearn.metrics import mean_squared_error
liner_model_mse=mean_squared_error(y_predict,ytest)
print("liner model_mse is ",math.sqrt(liner_model_mse))
```

## second algorithm

```python
from sklearn.linear_model import Lasso
lasso_model=Lasso(alpha=0.5)
lasso_model.fit(xtrain,ytrain)
```

```python
# check for the training accuracy
lasso_model.score(xtrain,ytrain)
```

```python
# checking for the testing accuracy
lasso_model.score(xtest,ytest)
```

```python
y_predict=lasso_model.predict(xtest)
lasso_model_mse=mean_squared_error(y_predict,ytest)
print("the square accuacy is ",math.sqrt(lasso_model_mse))
```

### third algorithm

```python
from sklearn.linear_model import Ridge
ridge_model=Ridge(alpha=0.2)
ridge_model.fit(xtrain,ytrain)
```

```python
#### traing accuracy
ridge_model.score(xtrain,ytrain)
```

```python
#### testig accuracy 
ridge_model.score(xtest,ytest)
```

```python
y_predict=ridge_model.predict(xtest)
# score=ridge_model
ridge_model_mse=mean_squared_error(y_predict,ytest)
print(" the mean square value is ",math.sqrt(ridge_model_mse))
```
